## Birth of a Signal: Signal Interactions and the Emergence of Pattern

**Date:** 2025-04-30

This notebook explores the idea that *life emerges not merely from material components, but from the interaction of rhythms and patterns* that animate them. Inspired by the conceptualization that two signals—each representing a parent waveform—can give rise to a third, emergent signal, we simulate various interaction modes between two sine waves.

The core question guiding this exploration is:

> What if life is fundamentally a rhythmic interaction between underlying signals—an emergent waveform that arises when dynamic entities temporarily entangle and resonate?

We define and visualize:
- The **father** and **mother** signals (pure sine waves of different frequencies)
- An **interaction envelope** that modulates how strongly these signals influence each other
- Three perspectives of the resulting **child signal**:
  - Father's perspective: `father + envelope × mother`
  - Mother's perspective: `mother + envelope × father`
  - Product-based perspective: `father × mother × envelope`

This experiment is part of a broader investigation into **rhythmic emergence**, **signal-based models of life**, and **process-based biology**.

Further notebooks will extend this idea to explore:
- Nonlinear signal coupling
- Pattern formation over space and time
- Stochastic resonance and noise-driven emergence
- Biological analogs in heartbeats, brain waves, and morphogenesis

```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling parameters
fs       = 1000       # Sampling frequency (Hz)
duration = 10.0       # Duration of the signal (seconds)
t        = np.linspace(0, duration, int(fs * duration), endpoint=False)

# Signal frequencies
f_father = 5.0        # Father frequency (Hz)
f_mother = 7.0        # Mother frequency (Hz)
f_mod    = 0.2        # Envelope (interaction) frequency (Hz)

# Generate the two parent signals
father = np.sin(2 * np.pi * f_father * t)
mother = np.sin(2 * np.pi * f_mother * t)

# Generate an envelope in [0, 1] that modulates interaction strength
envelope = 0.5 * (1 + np.sin(2 * np.pi * f_mod * t))

# Create three “child” signals from different perspectives:
child_father  = father + envelope * mother      # Father’s perspective
child_mother  = mother + envelope * father      # Mother’s perspective
child_product = father * mother * envelope      # Product perspective

# Plot pure parents and child signals
fig, axs = plt.subplots(5, 1, sharex=True, figsize=(12, 10))

# Pure father
axs[0].plot(t, father, color='tab:blue')
axs[0].set_title(f'Father Signal ({f_father} Hz)')
axs[0].set_ylabel('Amplitude')

# Pure mother
axs[1].plot(t, mother, color='tab:orange')
axs[1].set_title(f'Mother Signal ({f_mother} Hz)')
axs[1].set_ylabel('Amplitude')

# Child: Father's perspective
axs[2].plot(t, child_father, color='tab:green')
axs[2].set_title('Child (Father’s Perspective): father + envelope × mother')
axs[2].set_ylabel('Amplitude')

# Child: Mother's perspective
axs[3].plot(t, child_mother, color='tab:red')
axs[3].set_title('Child (Mother’s Perspective): mother + envelope × father')
axs[3].set_ylabel('Amplitude')

# Child: Product perspective
axs[4].plot(t, child_product, color='tab:purple')
axs[4].set_title('Child (Product Perspective): father × mother × envelope')
axs[4].set_ylabel('Amplitude')
axs[4].set_xlabel('Time [s]')

plt.tight_layout()
plt.show()
```
